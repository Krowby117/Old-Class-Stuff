```python
def count_moves(n: int, source: str, dest: str, via: str)->int:
    if n == 1: 
        return 1
    else: 
        return count_moves(n-1, source, via, dest) + 1 + count_moves(n-1, via, dest, source)
    
print(count_moves(3, "A", "B", "C"))
```

```python
def show_moves(n: int, A: str, B: str, C: str)->list:
    if n == 1: 
        return [(A, B)]
    return show_moves(n-1, A, C, B) + [(A, B)] + show_moves(n-1, C, B, A)
    
print(show_moves(3, "A", "B", "C"))
```
